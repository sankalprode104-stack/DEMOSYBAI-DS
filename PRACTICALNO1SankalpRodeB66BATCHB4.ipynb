Aim: Use sequence data types and their associated operations in Python.



Sequence data types in Python are ordered collections of items, meaning the order of elements is preserved. Each element can be accessed by its position (index).

**Common Sequence Types:**
*   **Strings (`str`):** Sequences of characters.
*   **Lists (`list`):** Mutable sequences of arbitrary objects.
*   **Tuples (`tuple`):** Immutable sequences of arbitrary objects.
*   **Bytes (`bytes`), Bytearrays (`bytearray`), Memoryviews (`memoryview`):** Sequences of bytes (less common for general programming).

**Indexing:**
Indexing allows you to access individual elements within a sequence. Python uses zero-based indexing, meaning the first element is at index 0, the second at index 1, and so on. Negative indices count from the end of the sequence, with -1 being the last element, -2 the second to last, etc.
*   `my_sequence[index]`
    *   Example: `my_string = "hello"; my_string[0]` returns `'h'`
    *   Example: `my_list = [1, 2, 3]; my_list[-1]` returns `3`

**Slicing:**
Slicing allows you to extract a sub-sequence (a 'slice') from a sequence. It uses a colon-separated notation `[start:stop:step]`:
*   `start`: The starting index (inclusive). If omitted, defaults to 0.
*   `stop`: The ending index (exclusive). If omitted, defaults to the end of the sequence.
*   `step`: The step or stride (how many items to skip). If omitted, defaults to 1.
    *   Example: `my_string = "Python"; my_string[1:4]` returns `'yth'` (elements at index 1, 2, 3)
    *   Example: `my_list = [1, 2, 3, 4, 5]; my_list[::2]` returns `[1, 3, 5]` (every second element)
    *   Example: `my_tuple = (10, 20, 30, 40); my_tuple[::-1]` returns `(40, 30, 20, 10)` (reverses the tuple)

Indexing and slicing work consistently across all standard sequence types (strings, lists, tuples).



**Differences:**
*   **Mutability:**
    *   **Lists are mutable:** Their elements can be changed, added, or removed after creation. They are defined using square brackets `[]`.
    *   **Tuples are immutable:** Once created, their elements cannot be changed, added, or removed. They are defined using parentheses `()`.
*   **Syntax:**
    *   Lists: `[1, 'apple', 3.14]`
    *   Tuples: `(1, 'apple', 3.14)`

**Why prefer a tuple over a list?**

1.  **Data Integrity/Immutability:** When you want to ensure that the data collected together won't be accidentally modified. For example, database records, coordinates (latitude, longitude), or RGB color values.
2.  **Performance:** Tuples are generally faster to iterate over and create than lists, especially for large collections, because their size and content are fixed.
3.  **Use as Dictionary Keys:** Since tuples are immutable, they can be used as keys in dictionaries, whereas lists cannot.
4.  **Function Arguments and Return Values:** Tuples are often used to return multiple values from a function, providing a simple way to group related data. They are also common for unpacking assignments.
5.  **Thread Safety:** In multi-threaded environments, immutable objects (like tuples) are inherently thread-safe because they cannot be changed by multiple threads simultaneously.



**Differences:**
*   **Element Type:**
    *   **Lists:** Can store elements of different data types (heterogeneous).
    *   **Arrays (from the `array` module):** Can only store elements of a single, specified numeric data type (homogeneous). This type is specified using a 'type code' when the array is created (e.g., `'i'` for signed int, `'f'` for float).
*   **Memory Usage:**
    *   **Lists:** Store references to objects, which can be scattered in memory. This leads to higher memory overhead.
    *   **Arrays:** Store actual values in a contiguous block of memory, making them much more memory-efficient for large collections of numbers.
*   **Performance:**
    *   **Lists:** Slower for numerical operations due to type checking and reference overhead.
    *   **Arrays:** Faster for numerical operations, especially when dealing with many numbers, because they are more compact and operations can be optimized at a lower level.

**When to use the `array` module instead of a list:**

*   **Memory Efficiency:** When you need to store a very large number of numeric values and memory is a concern. For example, storing millions of sensor readings or image pixel data.
*   **Performance with Numbers:** When you perform many numerical operations on a sequence of numbers and require better performance than a standard list can offer.
*   **Interfacing with C/C++:** When you need to interact with C or C++ code that expects a contiguous block of memory for numerical data.

For general-purpose collections of varied data types or when you don't have strict memory/performance requirements for numerical data, lists are usually preferred due to their flexibility and ease of use.



**How Dictionaries Work Internally:**

Python dictionaries are implemented as **hash tables**. Here's a simplified breakdown:

1.  **Hashing:** When you store a `(key, value)` pair, Python first calculates a **hash value** for the key. A hash function takes the key and produces an integer value. Good hash functions distribute keys evenly across a range of hash values.
2.  **Bucket/Slot Assignment:** The hash value is then used to determine a specific 'bucket' or 'slot' in an internal array where the `(key, value)` pair will be stored. This usually involves a modulo operation (`hash(key) % array_size`).
3.  **Collision Handling:** It's possible for two different keys to produce the same hash value or to map to the same bucket (a 'collision'). Python handles this, typically by storing multiple `(key, value)` pairs in the same bucket (e.g., using a linked list within each bucket) or by probing for the next available slot.
4.  **Equality Check:** When a lookup occurs, Python again hashes the provided key, goes to the corresponding bucket, and then iterates through the items in that bucket. For each item, it performs an **equality check** (`==`) to find the exact key before returning its associated value.

**Why Dictionary Lookups are Faster:**

Dictionary lookups are significantly faster (on average, **O(1) - constant time**) compared to lists or tuples because of the hashing mechanism:

*   **Direct Access vs. Sequential Search:**
    *   **Dictionaries:** Instead of searching through every item, the hash function directly guides Python to the likely location (bucket) of the key. This means that, ideally, regardless of how many items are in the dictionary, finding an item involves a fixed number of steps (calculate hash, go to bucket, possibly a few equality checks).
    *   **Lists/Tuples:** To find an element by value in a list or tuple, Python generally has to perform a **sequential search** (iterate from the beginning until the element is found). In the worst case, this means checking every element (**O(n) - linear time**), where `n` is the number of elements.

*   **Constant Time on Average:** While collisions can degrade performance to O(n) in worst-case scenarios (e.g., a very poorly designed hash function or malicious input), Python's hash table implementation is highly optimized to minimize collisions, making average lookup times very close to constant time.



**Differences Between Sets and Lists:**

*   **Order:**
    *   **Lists:** Are ordered collections. The elements maintain the order in which they were inserted.
    *   **Sets:** Are unordered collections. The elements do not maintain any specific order. (Note: Since Python 3.7+, `set` iteration order is insertion order, but this is an implementation detail and not part of the `set`'s core definition as an unordered mathematical set).
*   **Duplicates:**
    *   **Lists:** Can contain duplicate elements.
    *   **Sets:** Cannot contain duplicate elements. Each element in a set must be unique.
*   **Mutability of Elements:**
    *   **Lists:** Elements can be of any type.
    *   **Sets:** Elements must be hashable (immutable). This means you cannot put mutable objects like lists or dictionaries directly into a set.
*   **Syntax:**
    *   Lists: Defined using square brackets `[]`.
    *   Sets: Defined using curly braces `{}` or the `set()` constructor.

**How Sets Handle Duplicate Values:**

Sets inherently do not allow duplicates. If you try to add an element that already exists in a set, the set will simply ignore the new addition, and its content will remain unchanged. It effectively keeps only one instance of each unique element.

Example:
```python
my_list = [1, 2, 2, 3, 4, 4, 5]
my_set = set(my_list) # my_set will be {1, 2, 3, 4, 5}

my_set.add(3) # No change, 3 is already in the set
my_set.add(6) # Adds 6, my_set is now {1, 2, 3, 4, 5, 6}
```

**Where Sets Are Useful in Practical Scenarios:**

1.  **Removing Duplicates:** The most common use case is to easily get a collection of unique items from a list or other iterable. Just convert it to a set and back to a list if needed.
    *   Example: `unique_items = list(set(my_list))`
2.  **Membership Testing:** Checking if an element is present in a set is very fast (average O(1) time complexity), similar to dictionary lookups, because sets are also implemented using hash tables. This is much faster than checking membership in a list for large collections (O(n)).
    *   Example: `if 'apple' in my_set:`
3.  **Mathematical Set Operations:** Sets provide efficient methods for standard mathematical set operations:
    *   **Union (`|` or `union()`):** Elements in either set.
    *   **Intersection (`&` or `intersection()`):** Elements common to both sets.
    *   **Difference (`-` or `difference()`):** Elements in the first set but not in the second.
    *   **Symmetric Difference (`^` or `symmetric_difference()`):** Elements in either set, but not in both.
    *   Example: Finding common friends between two users.
4.  **Checking for Subsets/Supersets:** Efficiently determine if one set is a subset or superset of another (`issubset()`, `issuperset()`).

In essence, if you need a collection of unique, unordered items and frequently perform membership tests or set-theoretic operations, a set is the ideal data structure.
