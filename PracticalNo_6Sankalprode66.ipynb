## Answers to Web Scraping Questions

1.  **What is web scraping? Explain the role of the Requests library in web scraping.**
    *   **Web scraping** is an automated process of extracting information from websites. It involves programmatically fetching web pages and parsing their content to retrieve specific data. This data can then be stored, analyzed, or used for various applications.
    *   The **Requests library** in Python plays a crucial role in web scraping by handling the HTTP requests to web servers. It allows you to send GET, POST, PUT, DELETE, and other HTTP requests to retrieve the HTML content of a web page. Essentially, `Requests` acts like a web browser, fetching the raw source code of the web page that you then parse to extract data. It simplifies the process of making network requests and handling responses, managing things like cookies, sessions, and authentication.

2.  **What is BeautifulSoup? How does it help in parsing HTML?**
    *   **BeautifulSoup** (often referred to as `bs4`) is a Python library designed for parsing HTML and XML documents. It creates a parse tree from page source code that can be used to extract data in a structured and intuitive way.
    *   BeautifulSoup helps in parsing HTML by:
        *   **Providing Pythonic ways to navigate, search, and modify the parse tree:** You can easily find elements by tag name, ID, class, attributes, or even by text content.
        *   **Handling malformed HTML:** It's robust and can intelligently parse broken or poorly structured HTML, making it very practical for real-world web pages.
        *   **Abstraction:** It abstracts away the complexities of HTML structure, allowing developers to focus on the data they want to extract rather than intricate parsing logic.
        *   **Integration with parsers:** It works with popular Python parsers like `lxml` and `html.parser` to efficiently create its parse tree.

3.  **Differentiate between web scraping and web crawling.**
    *   **Web Scraping:** The act of extracting specific data from a particular web page or set of web pages. Its primary goal is data extraction. For example, getting all product prices from a specific e-commerce product category page.
    *   **Web Crawling:** The process of systematically browsing the World Wide Web, typically for the purpose of web indexing (as done by search engines). Its primary goal is to discover new URLs and build a map of the web. A web crawler (or spider) follows links from one page to another to discover and index content. Web scraping might be a component of web crawling, where data is extracted from the pages found by the crawler.

4.  **What are the commonly used methods in BeautifulSoup?**
    Some commonly used methods in BeautifulSoup include:
    *   `find()`: Finds the first tag that matches the given criteria.
    *   `find_all()`: Finds all tags that match the given criteria and returns them as a list.
    *   `.select()`: Uses CSS selectors to find elements, offering a powerful and concise way to locate tags.
    *   `.get_text()`: Extracts all the text content within a tag, stripping HTML tags.
    *   `.children`, `.contents`, `.parent`, `.next_sibling`, `.previous_sibling`: Methods for navigating the parse tree relative to a given tag.
    *   `['attribute_name']` (dictionary-style access): To get the value of an attribute of a tag (e.g., `<a>['href']` to get the URL).
    *   `.name`: To get the name of a tag.

5.  **What are the ethical and legal considerations in web scraping?**
    Web scraping involves several ethical and legal considerations:
    *   **Terms of Service (ToS):** Many websites explicitly prohibit scraping in their ToS. Violating these terms can lead to legal action, account suspension, or IP blocking.
    *   **Copyright:** Scraped data might be copyrighted. Reproducing, distributing, or using copyrighted content without permission can lead to infringement claims.
    *   **Privacy:** Scraping personal data (e.g., names, email addresses, phone numbers) without consent can violate privacy laws like GDPR or CCPA, leading to severe penalties.
    *   **Server Load:** Aggressive scraping can overload a website's server, potentially causing denial-of-service for legitimate users. This is both unethical and can be illegal.
    *   **Data Accuracy and Misrepresentation:** Presenting scraped data out of context or inaccurately can lead to misleading information and harm to individuals or businesses.
    *   **Robot Exclusion Protocol (robots.txt):** Websites often use `robots.txt` files to signal which parts of their site should not be crawled or scraped. While not legally binding in all jurisdictions, ignoring `robots.txt` is generally considered unethical.
    *   **Trespass to Chattels:** In some jurisdictions, accessing a computer system without authorization (e.g., circumventing anti-scraping measures) can be considered trespass to chattels, even if no damage is done.

    **Best practices include:** always checking a website's `robots.txt` and ToS, identifying yourself with a proper User-Agent, implementing delays between requests to avoid overloading servers, and only scraping publicly available data that is not subject to copyright or privacy concerns.

```python
# Importing required libraries
# requests → used to send HTTP requests to a website
# bs4 (BeautifulSoup) → used to parse HTML content
import requests
from bs4 import BeautifulSoup
# Step 1: Define the URL of the webpage you want to scrape
# Example: A simple static website for practice
url = "https://quotes.toscrape.com/"
# Step 2: Send an HTTP GET request to the URL
# This fetches the webpage content
response = requests.get(url)
# Step 3: Check if the request was successful
# Status code 200 means OK (successful request)
if response.status_code == 200:
    print("Successfully fetched the webpage")
else:
    print("Failed to fetch webpage")
    exit()  # Stop execution if request fails
# Step 4: Parse the HTML content using BeautifulSoup
# 'html.parser' is Python's built-in HTML parser
soup = BeautifulSoup(response.text, 'html.parser')
# Step 5: Find all quote containers on the page
# Each quote is inside a <div> with class 'quote'
quotes = soup.find_all('div', class_='quote')
# Step 6: Loop through each quote and extract data
for index, quote in enumerate(quotes):
  # Extract the quote text (inside <span class="text">)
    text = quote.find('span', class_='text').get_text()
# Extract the author name (inside <small class="author">)
    author = quote.find('small', class_='author').get_text()
# Extract tags (inside <a class="tag">)
    tags = quote.find_all('a', class_='tag')
# Convert list of tag elements into actual text list
    tag_list = [tag.get_text() for tag in tags]
# Print extracted data
    print(f"\nQuote {index + 1}:")
    print("Text:", text)
    print("Author:", author)
    print("Tags:", ", ".join(tag_list))
# Step 7: (Optional) Extract page title
title = soup.find('title').get_text()
print("\nPage Title:", title)
```
